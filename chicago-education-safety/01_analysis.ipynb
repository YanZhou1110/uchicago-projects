```python
import pandas as pd

df2012 = pd.read_csv('data/Crimes2012.csv')
df2013 = pd.read_csv('data/Crimes2013.csv')
df2014 = pd.read_csv('data/Crimes2014.csv')

df = pd.concat([df2012, df2013, df2014], ignore_index=True)

print(df.shape)
df.head()
```

```python
columns_needed = [
    "ID",
    "Date",
    "Primary Type",
    "Description",
    "Community Area",
    "Arrest",
    "Latitude",
    "Longitude"
]

df = df[columns_needed]
```

```python
df["Date"] = df["Date"].astype(str)
df["Date"] = df["Date"].str.replace(r"\s+", " ", regex=True).str.strip()

df["Date"] = pd.to_datetime(
    df["Date"],
    format="%m/%d/%Y %I:%M:%S %p",
    errors="coerce"
)

df = df.dropna(subset=["Date"])

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
```

```python
df = df.dropna(subset=["Community Area"])

df = df[df["Community Area"] != 0]

df = df.drop_duplicates()
```

```python
yearly = df.groupby("Year").size()

print(yearly)

yearly.plot(kind="line", marker="o", figsize=(8,5))
```

```python
df = df[df["Community Area"].notna()]
df = df[df["Community Area"] != 0]
df["Community Area"] = df["Community Area"].astype(int)

area_counts = df.groupby("Community Area").size().sort_values(ascending=False)

top5 = area_counts.head(5)
safe5 = area_counts.tail(5)

print(top5)
print(safe5)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2, 1, figsize=(10,8))

top5.sort_values().plot(kind="barh", ax=ax[0])
ax[0].set_title("Top 5 Most Dangerous Neighborhoods (Community Areas)")
ax[0].set_xlabel("Crime Count")
ax[0].set_ylabel("Community Area")

safe5.sort_values().plot(kind="barh", ax=ax[1])
ax[1].set_title("Top 5 Safest Neighborhoods (Community Areas)")
ax[1].set_xlabel("Crime Count")
ax[1].set_ylabel("Community Area")

plt.tight_layout()
```

```python
import matplotlib.pyplot as plt

crime_type = df["Primary Type"].value_counts().head(10)

crime_type.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 Crime Types in Chicago (2012-2014)")
plt.xlabel("Crime Type")
plt.ylabel("Count")
```

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month

monthly_year = df.groupby(["Month","Year"]).size().unstack()

monthly_year = monthly_year.reindex(range(1,13))

monthly_year.plot(figsize=(10,5), marker="o")

plt.title("Monthly Crime Trend by Year (2012–2014)")
plt.xlabel("Month")
plt.ylabel("Crime Count")
```
